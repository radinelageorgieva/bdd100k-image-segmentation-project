## BDD100K Image Segmentation Project

This is a computer vision project focused on semantic image segmentation using the BDD100K dataset. The goal is to train a model that accurately identifies drivable areas, lanes, and traffic objects to support autonomous driving systems.

```python
data_path = r"C:\Users\Petar\Desktop\New folder\Deep learning\Segmentation\bdd100k"
```

```python
#generate paths for train, val, and test splits
splits = ["train", "val", "test"]
```

```python
paths = {
    split: {
        "images": os.path.join(data_path, "images", "100k", split),
        "labels": os.path.join(data_path, "labels", split)
    }
    for split in splits
}
```

```python
print(paths['test']['images'])
print(paths['test']['labels'])
```

```python
#create custom dataset class to handle the loading pipelinе
class BDD100KDataset(Dataset):
    def __init__(self, images_dir, labels_dir):
        self.images_dir = images_dir
        self.labels_dir = labels_dir
    
        self.image_names = sorted(os.listdir(images_dir))

    def __len__(self):
        return len(self.image_names)

    def __getitem__(self, idx):
        img_name = self.image_names[idx]
        img_path = os.path.join(self.images_dir, img_name)
        
        #convert image to RGB to standardize channels
        image = Image.open(img_path).convert("RGB")
        
        return image
```

```python
#initializa datasets
datasets = {}

#images and labels preprocesing for training
for split in splits:
    datasets[split] = BDD100KDataset(
        images_dir = paths[split]["images"],
        labels_dir = paths[split]["labels"],
    
    )
```

```python
print(f"Number of train images: {len(datasets['train'])}")
print(f"Number of val images: {len(datasets['val'])}")
print(f"Number of test images: {len(datasets['test'])}")
```
